```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense
model = Sequential() # 建立堆疊模型
model.add(Dense(32, activation='relu', input_dim=10)) #建立第一層，有32個神經元，輸入維度為10，啟動函數為relu
model.add(Dense(1, activation='sigmoid')) #建立第二層，有1個神經元(剛好為輸出結果維度，啟動函數維relu
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy']) #編譯模型，採用rmsprop優化器，損失函數為binary_crossentropy，比對結果好壞方式為accuracy正確性
import numpy as np
data = np.random.random((1000, 10)) #隨機產生1000筆資料，每筆資料包含10個float數字
labels = np.random.randint(2, size=(1000, 1)) #隨機產生1000筆答案，每筆資料上限為2(意思就是0 或 1)
print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])
model.fit(data, labels, epochs=10, batch_size=32) #丟資料進去訓練
score=model.evaluate(data, labels) #評估正確率
print(score[1]) #score[0]為總誤差，score[1]為正確率
```

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Dropout
model = Sequential() # 建立堆疊模型
model.add(Dense(512, activation='relu', input_dim=784)) #建立第一層，有512個神經元，輸入維度為784，啟動函數為relu
model.add(Dropout(0.2)) #隨機丟棄20%的參數
model.add(Dense(10, activation='softmax')) #建立第二層，有10個神經元(剛好為輸出結果維度，啟動函數維softmax
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']) #編譯模型，採用adam優化器，損失函數為categorial_crossentropy，比對結果好壞方式為accuracy正確性
import numpy as np
data = np.random.random((1000, 784)) #隨機產生1000筆資料，每筆資料包含784個float數字
labels = np.random.randint(10, size=(1000, 10))#隨機產生1000筆答案，每筆資料上限為10(意思就是0~9)

model.fit(data, labels, epochs=10, batch_size=32) #丟資料進去訓練
score=model.evaluate(data, labels) #評估正確率
print(score[1]) #score[0]為總誤差，score[1]為正確率
```

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np
data = np.random.random((1000, 10)) #隨機產生1000筆資料，每筆資料包含十個float數字
labels = np.random.randint(2, size=(1000, 1)) #隨機產生1000筆資料，每筆資料上限為2(意思就是0 或 1)
print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])

model = Sequential()
model.add(Dense(32, activation='relu', input_dim=10))
model.add(Dense(1, activation='sigmoid'))
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

model.fit(data, labels, epochs=10, batch_size=32)

score=model.evaluate(data, labels)
print(score[1])
```
